```python
number1 = int(input("Inserta el numero 1 : "))
number2 = int(input("Inserta el numero 2 : "))

def sum(number1, number2):
    result = number1 + number2
    print(number1, "+", number2, "=", result)

def sub(number1, number2):
    result = number1 - number2
    print(number1, "-", number2, "=", result)

def mult(number1, number2):
    result = number1 * number2
    print(number1, "*", number2, "=", result)

def div(number1, number2):
    if (number2 != 0):
       result = number1 / number2
       print(number1, "/", number2, "=", result)
    else:
        print("El numero 2 es igual a 0 y no se puede dividir por 0")

sum(number1, number2)
sub(number1, number2)
mult(number1, number2)
div(number1, number2)
```

```python
# Solicita al usuario el primer número mediante un texto en consola,
# convierte la entrada (que por defecto es texto) a un número entero y lo guarda en la variable 'number1'.
number1 = int(input("Inserta el numero 1 : "))

# Solicita al usuario el segundo número, lo convierte a un entero
# y lo almacena en la variable 'number2'.
number2 = int(input("Inserta el numero 2 : "))


# Define la función 'sum' que recibe dos parámetros (number1 y number2) para realizar una suma.
def sum(number1, number2):
    # Suma los dos valores recibidos y guarda el resultado en la variable local 'result'.
    result = number1 + number2
    # Imprime en consola la operación matemática realizada y su resultado (ej. 5 + 3 = 8).
    print(number1, "+", number2, "=", result)


# Define la función 'sub' que recibe dos parámetros para realizar una resta.
def sub(number1, number2):
    # Resta el segundo número al primero y almacena el resultado en 'result'.
    result = number1 - number2
    # Muestra en consola la operación de resta y su resultado.
    print(number1, "-", number2, "=", result)


# Define la función 'mult' que recibe dos parámetros para realizar una multiplicación.
def mult(number1, number2):
    # Multiplica ambos números y guarda el producto en la variable 'result'.
    result = number1 * number2
    # Muestra en pantalla el formato de la multiplicación junto a su resultado.
    print(number1, "*", number2, "=", result)


# Define la función 'div' que recibe dos parámetros para realizar una división.
def div(number1, number2):
    # Comprueba si el segundo número es distinto de cero para evitar un error de división por cero.
    if number2 != 0:
        # Si la condición se cumple, divide el primer número entre el segundo y guarda el resultado.
        result = number1 / number2
        # Imprime la operación de división realizada y su resultado.
        print(number1, "/", number2, "=", result)
    # Si el segundo número es cero, se ejecuta esta parte del código.
    else:
        # Muestra un mensaje de advertencia indicando que no es posible dividir entre cero.
        print("El numero 2 es igual a 0 y no se puede dividir por 0")


# Llama a la función 'sum' pasándole los datos introducidos por el usuario para ejecutar la suma.
sum(number1, number2)

# Llama a la función 'sub' pasándole los mismos datos para ejecutar la resta.
sub(number1, number2)

# Llama a la función 'mult' para calcular y mostrar la multiplicación.
mult(number1, number2)

# Llama a la función 'div' para evaluar y ejecutar la división si es válida.
div(number1, number2)
```
